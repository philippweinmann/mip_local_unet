Let's create a minimal viable graph example.

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import json
```

```python
# let's generate a dummy, undirected graph

def generate_graph():
    G = nx.Graph()
    # artery 1
    G.add_edge(1, 2)
    G.add_edge(2, 3)
    G.add_edge(3, 4)
    G.add_edge(4, 5)

    # artery 2
    G.add_edge(6, 7)
    G.add_edge(7, 8)

    for node in G.nodes:
        G.nodes[node]['is_root'] = False
        G.nodes[node]['pos'] = list(np.random.rand(3))

    for edge in G.edges:
        G.edges[edge]['length'] = float(np.random.rand()*10)

    G.nodes[1]['is_root'] = True
    G.nodes[6]['is_root'] = True

    return G


g = generate_graph()
```

```python
# another graph but with minimally changed positions and lengths

# let's get the first graph in json format
data1 = nx.node_link_data(g, edges = "edges")

g2 = g.copy()

for node in g2.nodes:
    g2.nodes[node]['pos'] += list(np.random.rand(3))

for edge in g2.edges:
    print(g2.edges[edge]['length'])
    g2.edges[edge]['length'] += float(np.random.rand())
    print(g2.edges[edge]['length'])

data2 = nx.node_link_data(g2, edges = "edges")
```

```python
# let's save the json data to a file
label_folder = "graph_label"
pred_folder = "graph_test"

dummy_idxs = [] 

for i in range(10):
    dummy_idxs.append("abcd" + str(i)) 

for dummy_idx in dummy_idxs:
    with open(f"{label_folder}/{dummy_idx}.graph.json", "w") as f:
        json.dump(data1, f)

    with open(f"{pred_folder}/{dummy_idx}.graph.json", "w") as f:
        json.dump(data2, f)
```
